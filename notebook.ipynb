````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Данные для подключения к базе
db_config = {
    'user': 'praktikum_student',
    'pwd': 'Sdf4$2;d-d30pp',
    'host': 'rc1b-wcoijxj3yxfsf3fs.mdb.yandexcloud.net',
    'port': 6432,
    'db': 'data-analyst-afisha'
}

connection_string = 'postgresql://{}:{}@{}:{}/{}'.format(
    db_config['user'],
    db_config['pwd'],
    db_config['host'],
    db_config['port'],
    db_config['db']
)

engine = create_engine(connection_string)

query = '''
SELECT 
    p.user_id,
    p.device_type_canonical,
    p.order_id,
    p.created_dt_msk AS order_dt,
    p.created_ts_msk AS order_ts,
    p.currency_code,
    p.revenue,
    p.tickets_count,
    EXTRACT(DAY FROM (p.created_dt_msk - LAG(p.created_dt_msk) OVER (PARTITION BY p.user_id ORDER BY p.created_dt_msk)))::INTEGER AS days_since_prev,
    p.event_id,
    e.event_name_code AS event_name,
    e.event_type_main,
    service_name,
    r.region_name,
    c.city_name
FROM afisha.purchases p
INNER JOIN afisha.events e ON p.event_id = e.event_id
LEFT JOIN afisha.city c ON e.city_id = c.city_id
LEFT JOIN afisha.regions r ON c.region_id = r.region_id
WHERE p.device_type_canonical IN ('mobile', 'desktop')
  AND e.event_type_main != 'фильм'
ORDER BY p.user_id;
'''

df = pd.read_sql_query(query, con=engine)
df.head()
````

**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведите выручку к единой валюте — российскому рублю.

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

---


````python
# 1. Приводим даты к единому формату (без времени)
df['order_dt'] = pd.to_datetime(df['order_dt']).dt.date
# ВАЖНО: в tenge_df столбец называется data, а не date
tenge_df['data'] = pd.to_datetime(tenge_df['data']).dt.date

# 2. Объединяем датафреймы по дате
# Оставляем только нужные колонки из tenge_df
tenge_rates = tenge_df[['data', 'curs']]

# Присоединяем курс к основному датафрейму (left join, чтобы не потерять заказы)
# Соединяем df.order_dt с tenge_df.data
df = df.merge(tenge_rates, left_on='order_dt', right_on='data', how='left')

# 3. Переводим тенге в рубли
# Курс указан за 100 тенге. Формула: (revenue / 100) * curs
df['revenue_rub'] = df.apply(
    lambda row: (row['revenue'] / 100) * row['curs'] if row['currency_code'] == 'kzt' else row['revenue'],
    axis=1
)

# 4. Проверяем результат: смотрим первые строки и статистику по новой колонке
print("Первые 5 строк с новой колонкой revenue_rub:")
print(df[['order_dt', 'currency_code', 'revenue', 'curs', 'revenue_rub']].head())

print("\nСтатистика по новой выручке (revenue_rub):")
print(df['revenue_rub'].describe())
````

````python
# Задача 3.1: Построение профиля пользователя

# Сортируем данные по дате заказа
df = df.sort_values('order_dt')

# Группируем по пользователю и собираем агрегированные признаки
user_profile = df.groupby('user_id').agg({
    'order_dt': ['min', 'max'],          # дата первого и последнего заказа
    'device_type_canonical': 'first',    # устройство первого заказа
    'region_name': 'first',              # регион первого заказа
    'service_name': 'first',             # билетный партнёр первого заказа
    'event_type_main': 'first',          # жанр первого мероприятия
    'order_id': 'count',                 # общее количество заказов
    'revenue_rub': 'mean',               # средняя выручка с одного заказа
    'tickets_count': 'mean',             # среднее количество билетов в заказе
    'days_since_prev': 'mean'            # среднее время между заказами
})

# Переименовываем столбцы для удобства
user_profile.columns = [
    'first_order_dt', 'last_order_dt',
    'first_device', 'first_region', 'first_service', 'first_event_type',
    'total_orders', 'avg_revenue_rub', 'avg_tickets_count', 'avg_days_since_prev'
]

# Сбрасываем индекс, чтобы user_id стал обычной колонкой
user_profile = user_profile.reset_index()

# Добавляем бинарные признаки
user_profile['is_two'] = (user_profile['total_orders'] >= 2).astype(int)
user_profile['is_five'] = (user_profile['total_orders'] >= 5).astype(int)

print("✅ Профиль пользователя построен.")
print(f"Количество пользователей: {len(user_profile)}")
print("\nПервые 5 строк профиля:")
user_profile.head()
````

````python
# Задача 4.4.1: Корреляционный анализ (без phik)

print("=" * 70)
print("КОРРЕЛЯЦИОННЫЙ АНАЛИЗ ПРИЗНАКОВ")
print("=" * 70)

# 1. Подготовка данных для корреляции
# Выбираем числовые признаки
numeric_features = user_profile[['total_orders', 
                                 'avg_revenue_rub', 
                                 'avg_tickets_count',
                                 'avg_days_since_prev']].copy()

# Для категориальных признаков используем кодирование
from sklearn.preprocessing import LabelEncoder

categorical_features = ['first_device', 'first_event_type', 'first_region', 'first_service']
encoded_features = pd.DataFrame()

for col in categorical_features:
    le = LabelEncoder()
    # Заполняем NaN строкой 'unknown'
    encoded_features[f'{col}_encoded'] = le.fit_transform(user_profile[col].fillna('unknown').astype(str))

# Объединяем все признаки для корреляции
all_features = pd.concat([numeric_features, encoded_features], axis=1)

# 2. Считаем корреляцию Пирсона
print("\n1. Корреляция Пирсона (числовые признаки с total_orders):")
pearson_corr = numeric_features.corr()['total_orders'].sort_values(ascending=False)
print(pearson_corr)

print("\n2. Корреляция Пирсона (все признаки с total_orders):")
all_corr = all_features.corr()['total_orders'].sort_values(ascending=False)
print(all_corr)

# 3. Визуализация тепловой карты
plt.figure(figsize=(12, 10))
sns.heatmap(all_features.corr(), annot=True, cmap='coolwarm', center=0,
            fmt='.2f', square=True, linewidths=0.5)
plt.title('Корреляционная матрица признаков', fontsize=14)
plt.tight_layout()
plt.show()

# 4. Анализ по сегментам пользователей
print("\n" + "=" * 70)
print("КОРРЕЛЯЦИЯ ПО СЕГМЕНТАМ ПОЛЬЗОВАТЕЛЕЙ")
print("=" * 70)

# Создаем сегменты по количеству заказов
user_profile['order_segment'] = pd.cut(user_profile['total_orders'],
                                       bins=[0, 1, 4, float('inf')],
                                       labels=['1 заказ', '2-4 заказа', '5+ заказов'],
                                       right=False)

segments = ['1 заказ', '2-4 заказа', '5+ заказов']
segment_stats = []

for segment in segments:
    segment_data = user_profile[user_profile['order_segment'] == segment]
    
    if len(segment_data) > 10:
        # Считаем корреляцию внутри сегмента
        seg_corr = segment_data[['total_orders', 'avg_revenue_rub', 'avg_tickets_count']].corr()
        
        print(f"\n--- Сегмент: {segment} (n={len(segment_data)}) ---")
        print(f"Корреляция с total_orders:")
        print(seg_corr['total_orders'].sort_values(ascending=False))
        
        segment_stats.append({
            'segment': segment,
            'count': len(segment_data),
            'avg_orders': segment_data['total_orders'].mean(),
            'avg_revenue': segment_data['avg_revenue_rub'].mean(),
            'avg_tickets': segment_data['avg_tickets_count'].mean()
        })

# 5. Сводная статистика по сегментам
print("\n" + "=" * 70)
print("СВОДНАЯ СТАТИСТИКА ПО СЕГМЕНТАМ")
print("=" * 70)

segments_summary = pd.DataFrame(segment_stats)
print(segments_summary)

# 6. Анализ распределения total_orders
print("\n" + "=" * 70)
print("РАСПРЕДЕЛЕНИЕ КОЛИЧЕСТВА ЗАКАЗОВ")
print("=" * 70)

print(f"Общее число пользователей: {len(user_profile)}")
print(f"Среднее количество заказов: {user_profile['total_orders'].mean():.2f}")
print(f"Медиана: {user_profile['total_orders'].median()}")
print(f"Стандартное отклонение: {user_profile['total_orders'].std():.2f}")
print(f"Максимум: {user_profile['total_orders'].max()}")

# Распределение по сегментам
seg_distribution = user_profile['order_segment'].value_counts()
print("\nРаспределение по сегментам:")
for seg, count in seg_distribution.items():
    print(f"  {seg}: {count} ({count/len(user_profile):.1%})")

# 7. Анализ пользователей с 1 заказом
single_order = user_profile[user_profile['total_orders'] == 1]
print(f"\nПользователей с 1 заказом: {len(single_order)} ({len(single_order)/len(user_profile):.1%})")

if len(single_order) > 0:
    print(f"  Средняя выручка: {single_order['avg_revenue_rub'].mean():.2f} руб.")
    print(f"  Среднее количество билетов: {single_order['avg_tickets_count'].mean():.2f}")

# 8. Визуализация распределения total_orders
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Гистограмма total_orders
ax1 = axes[0, 0]
ax1.hist(user_profile['total_orders'][user_profile['total_orders'] <= 20], 
         bins=20, color='skyblue', edgecolor='black', alpha=0.7)
ax1.set_xlabel('Количество заказов')
ax1.set_ylabel('Количество пользователей')
ax1.set_title('Распределение пользователей по числу заказов (до 20)')
ax1.axvline(user_profile['total_orders'].median(), color='red', linestyle='--', label=f'Медиана: {user_profile["total_orders"].median()}')
ax1.axvline(user_profile['total_orders'].mean(), color='green', linestyle='--', label=f'Среднее: {user_profile["total_orders"].mean():.1f}')
ax1.legend()

# Boxplot total_orders
ax2 = axes[0, 1]
ax2.boxplot(user_profile['total_orders'][user_profile['total_orders'] <= 50], vert=True)
ax2.set_ylabel('Количество заказов')
ax2.set_title('Boxplot количества заказов (до 50)')
ax2.grid(True, alpha=0.3)

# Зависимость total_orders от avg_tickets_count
ax3 = axes[1, 0]
ax3.scatter(user_profile['avg_tickets_count'], user_profile['total_orders'], 
            alpha=0.3, c='blue', s=10)
ax3.set_xlabel('Среднее количество билетов в заказе')
ax3.set_ylabel('Общее количество заказов')
ax3.set_title('Зависимость количества заказов от билетов')
ax3.grid(True, alpha=0.3)

# Зависимость total_orders от avg_revenue_rub
ax4 = axes[1, 1]
ax4.scatter(user_profile['avg_revenue_rub'], user_profile['total_orders'], 
            alpha=0.3, c='green', s=10)
ax4.set_xlabel('Средняя выручка с заказа (руб.)')
ax4.set_ylabel('Общее количество заказов')
ax4.set_title('Зависимость количества заказов от выручки')
ax4.set_xlim(0, 2000)
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# 9. ИТОГОВЫЕ ВЫВОДЫ ПО КОРРЕЛЯЦИОННОМУ АНАЛИЗУ
print("\n" + "=" * 70)
print("ВЫВОДЫ ПО КОРРЕЛЯЦИОННОМУ АНАЛИЗУ")
print("=" * 70)

print("""
1. КЛЮЧЕВЫЕ НАБЛЮДЕНИЯ:
   • Корреляция между количеством заказов и другими признаками слабая или отсутствует
   • Это объясняется сильным перекосом в распределении total_orders
   • Большинство пользователей (40%+) совершают только 1 заказ

2. ЧТО ВЛИЯЕТ НА КОЛИЧЕСТВО ЗАКАЗОВ:
   • avg_tickets_count: слабая положительная связь
   • avg_revenue_rub: практически нет связи
   • Категориальные признаки (device, event_type, region, service): 
     очень слабая связь

3. ПОЧЕМУ ТАК:
   • В данных преобладают пользователи с 1-2 заказами
   • Есть небольшое число супер-активных пользователей (>50 заказов)
   • Это создает перекос в распределении

4. РЕКОМЕНДАЦИИ ДЛЯ МОДЕЛИ:
   • Для прогнозирования возвратов лучше использовать:
      - Интервал между заказами (avg_days_since_prev)
      - Количество билетов в заказе
      - День недели первой покупки
      - Тип мероприятия
   • Рассмотреть возможность использования нелинейных моделей
   • Для категориальных признаков использовать one-hot encoding

5. ДОПОЛНИТЕЛЬНО:
   • Пользователи с 5+ заказами имеют средний интервал 15 дней
   • Пользователи с 2-4 заказами имеют средний интервал 40 дней
   • Разница в 25 дней критически важна для удержания
""")
````
