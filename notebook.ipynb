````python
import os

# Временно задаем переменные, чтобы код работал в тренажере
os.environ['DB_USER'] = 'praktikum_student'
os.environ['DB_PASSWORD'] = 'Sdf4$2;d-d30pp'
os.environ['DB_HOST'] = 'rc1b-wcoijxj3yxfsf3fs.mdb.yandexcloud.net'
os.environ['DB_PORT'] = '6432'
os.environ['DB_NAME'] = 'data-analyst-afisha'

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Данные для подключения к базе (загружаются из переменных окружения)
db_config = {
    'user': os.getenv('DB_USER'),
    'pwd': os.getenv('DB_PASSWORD'),
    'host': os.getenv('DB_HOST'),
    'port': os.getenv('DB_PORT'),
    'db': os.getenv('DB_NAME')
}

# Исправленная строка подключения (f-строка, не выдаст KeyError)
connection_string = f"postgresql://{db_config['user']}:{db_config['pwd']}@{db_config['host']}:{db_config['port']}/{db_config['db']}"

engine = create_engine(connection_string)

query = '''
SELECT 
    p.user_id,
    p.device_type_canonical,
    p.order_id,
    p.created_dt_msk AS order_dt,
    p.created_ts_msk AS order_ts,
    p.currency_code,
    p.revenue,
    p.tickets_count,
    EXTRACT(DAY FROM (p.created_dt_msk - LAG(p.created_dt_msk) OVER (PARTITION BY p.user_id ORDER BY p.created_dt_msk)))::INTEGER AS days_since_prev,
    p.event_id,
    e.event_name_code AS event_name,
    e.event_type_main,
    service_name,
    r.region_name,
    c.city_name
FROM afisha.purchases p
INNER JOIN afisha.events e ON p.event_id = e.event_id
LEFT JOIN afisha.city c ON e.city_id = c.city_id
LEFT JOIN afisha.regions r ON c.region_id = r.region_id
WHERE p.device_type_canonical IN ('mobile', 'desktop')
  AND e.event_type_main != 'фильм'
ORDER BY p.user_id;
'''

df = pd.read_sql_query(query, con=engine)
df.head()
````

**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведите выручку к единой валюте — российскому рублю.

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

---


````python
# 1. Приводим даты к единому формату (без времени)
df['order_dt'] = pd.to_datetime(df['order_dt']).dt.date
# ВАЖНО: в tenge_df столбец называется data, а не date
tenge_df['data'] = pd.to_datetime(tenge_df['data']).dt.date

# 2. Объединяем датафреймы по дате
# Оставляем только нужные колонки из tenge_df
tenge_rates = tenge_df[['data', 'curs']]

# Присоединяем курс к основному датафрейму (left join, чтобы не потерять заказы)
# Соединяем df.order_dt с tenge_df.data
df = df.merge(tenge_rates, left_on='order_dt', right_on='data', how='left')

# 3. Переводим тенге в рубли
# Курс указан за 100 тенге. Формула: (revenue / 100) * curs
df['revenue_rub'] = df.apply(
    lambda row: (row['revenue'] / 100) * row['curs'] if row['currency_code'] == 'kzt' else row['revenue'],
    axis=1
)

# 4. Проверяем результат: смотрим первые строки и статистику по новой колонке
print("Первые 5 строк с новой колонкой revenue_rub:")
print(df[['order_dt', 'currency_code', 'revenue', 'curs', 'revenue_rub']].head())

print("\nСтатистика по новой выручке (revenue_rub):")
print(df['revenue_rub'].describe())
````

### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас они просят вас построить агрегированные признаки, описывающие поведение и профиль каждого пользователя.

---

**Задача 3.1.** Постройте профиль пользователя — для каждого пользователя найдите:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавьте два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.

**Рекомендация:** перед тем как строить профиль, отсортируйте данные по времени совершения заказа.

---


````python
# Задача 3.1: Построение профиля пользователя

# Сортируем данные по дате заказа
df = df.sort_values('order_dt')

# Группируем по пользователю и собираем агрегированные признаки
user_profile = df.groupby('user_id').agg({
    'order_dt': ['min', 'max'],          # дата первого и последнего заказа
    'device_type_canonical': 'first',    # устройство первого заказа
    'region_name': 'first',              # регион первого заказа
    'service_name': 'first',             # билетный партнёр первого заказа
    'event_type_main': 'first',          # жанр первого мероприятия
    'order_id': 'count',                 # общее количество заказов
    'revenue_rub': 'mean',               # средняя выручка с одного заказа
    'tickets_count': 'mean',             # среднее количество билетов в заказе
    'days_since_prev': 'mean'            # среднее время между заказами
})

# Переименовываем столбцы для удобства
user_profile.columns = [
    'first_order_dt', 'last_order_dt',
    'first_device', 'first_region', 'first_service', 'first_event_type',
    'total_orders', 'avg_revenue_rub', 'avg_tickets_count', 'avg_days_since_prev'
]

# Сбрасываем индекс, чтобы user_id стал обычной колонкой
user_profile = user_profile.reset_index()

# Добавляем бинарные признаки
user_profile['is_two'] = (user_profile['total_orders'] >= 2).astype(int)
user_profile['is_five'] = (user_profile['total_orders'] >= 5).astype(int)

print("✅ Профиль пользователя построен.")
print(f"Количество пользователей: {len(user_profile)}")
print("\nПервые 5 строк профиля:")
user_profile.head()
````

````python
# Задача 4.3.1: Анализ влияния дня недели первой покупки

# 1. Сначала проверяем тип данных и преобразуем в datetime, если нужно
print("Тип данных first_order_dt:", user_profile['first_order_dt'].dtype)

# Если это не datetime, конвертируем
if user_profile['first_order_dt'].dtype != 'datetime64[ns]':
    user_profile['first_order_dt'] = pd.to_datetime(user_profile['first_order_dt'])
    print("Колонка преобразована в datetime")

# 2. Выделяем день недели из даты первого заказа
user_profile['first_weekday'] = user_profile['first_order_dt'].dt.dayofweek

# 3. Заменяем цифры на названия дней
weekday_names = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']
user_profile['first_weekday_name'] = user_profile['first_weekday'].map(lambda x: weekday_names[x])

# 4. Считаем количество пользователей и долю возвратов по дням недели
weekday_stats = user_profile.groupby('first_weekday_name').agg({
    'user_id': 'count',
    'is_two': 'mean'
}).rename(columns={'user_id': 'total_users', 'is_two': 'return_rate'})

# 5. Сортируем по порядку дней недели
weekday_order = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']
weekday_stats = weekday_stats.reindex(weekday_order)

print("=" * 60)
print("СТАТИСТИКА ПО ДНЯМ НЕДЕЛИ")
print("=" * 60)
print(weekday_stats)

# 6. Визуализация
fig, ax1 = plt.subplots(figsize=(10, 6))

bars = ax1.bar(weekday_stats.index, weekday_stats['total_users'], 
               color='lightgreen', alpha=0.7, label='Количество пользователей')
ax1.set_xlabel('День недели', fontsize=12)
ax1.set_ylabel('Количество пользователей', fontsize=12)

ax2 = ax1.twinx()
line = ax2.plot(weekday_stats.index, weekday_stats['return_rate'], 
                color='orange', marker='o', linewidth=2, label='Доля возвратов')
ax2.set_ylabel('Доля возвратов', fontsize=12)
ax2.set_ylim(0, 1)

lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines + lines2, labels + labels2, loc='upper left')

plt.title('Влияние дня недели первой покупки на возврат', fontsize=14)
plt.tight_layout()
plt.show()
````
